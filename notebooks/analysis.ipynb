# golden-vortex -- Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## Data Loading


```python
df = pd.read_csv('../data/sample.csv')
df.head()
```

## Summary Statistics


```python
df.describe()
```

## Visualization


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.histplot(df['value'], kde=True, ax=axes[0])
df.plot(x='timestamp', y='value', ax=axes[1])
plt.tight_layout()
plt.show()
```
